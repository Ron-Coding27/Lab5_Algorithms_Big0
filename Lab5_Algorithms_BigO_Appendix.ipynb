```python
LAST_NAME = "Sagcal"
STUDENT_ID = "TUPM-26-1864"
FAVORITE_ARTIST = "Arthur Nery"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha())

print("Seed:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
print("Artist Length:", ARTIST_LENGTH)
print("Artist Sum:", ARTIST_SUM)
```

```python
# [Task 1] Emergency Dispatch Record Search
original_dataset = [11, 22, 33, 44, 55]
rotated_dataset = original_dataset[SEED_DIGIT:] + original_dataset[:SEED_DIGIT]
target_incident = (ID_SUM + ARTIST_SUM) % 100

li, lc = linear_search(rotated_dataset, target_incident)
bi, bc = binary_search(rotated_dataset, target_incident)

print("Original Dataset:", original_dataset)
print("Rotated Dataset:", rotated_dataset)
print("Target Incident:", target_incident)
print("Binary Search Comparisons:", bc)
print("Linear Search Comparisons:", lc)
print("Time Complexity: Binary O(log n), Linear O(n)")
print("Space Complexity: O(1)")
print("Final Output:", "Found at index" if bi != -1 else "Not Found")
```

```python
# [Task 2] University Enrollment Records
deptA = [SEED_DIGIT * i for i in range(1, 6)]
deptB = [ARTIST_LENGTH * i for i in range(1, 5)]
consolidated = sorted(deptA + deptB)

print("\nOUTPUT 2: University Enrollment Records")
print("Inputs -> Seed Digit:", SEED_DIGIT, "Artist Length:", ARTIST_LENGTH, "Name Length:", NAME_LENGTH)
print("Department A Dataset:", deptA)
print("Department B Dataset:", deptB)
print("Consolidated Dataset:", consolidated)
print("Number of Records Before:", len(deptA) + len(deptB))
print("Number of Records After:", len(consolidated))
print("Comparison Count:", len(consolidated) - 1)
print("Ordering Verification:", consolidated == sorted(consolidated))
print("Time Complexity: O(n+m)")
print("Space Complexity: O(n+m)")
print("Final Output: Consolidated enrollment list generated successfully")
```
